# Student Attendance Analysis (20 Students)

This notebook uses a sample dataset of 20 students to calculate attendance percentages, identify students with low attendance, and visualize the results.

**Note:** The student names and attendance figures are fictional sample data. The dataset assumes 60 working days.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample attendance data for 20 students (out of 60 working days)
data = {
    'Student_ID': [f'S{i:02d}' for i in range(1, 21)],
    'Student_Name': [
        'Aarav', 'Aditi', 'Rohan', 'Priya', 'Karan',
        'Neha', 'Arjun', 'Isha', 'Rahul', 'Sneha',
        'Vikram', 'Ananya', 'Aditya', 'Meera', 'Kabir',
        'Pooja', 'Dev', 'Simran', 'Yash', 'Tanya'
    ],
    'Present_Days': [55, 52, 48, 58, 45, 50, 57, 54, 42, 56,
                     49, 59, 51, 47, 53, 44, 60, 46, 55, 50],
    'Total_Days': [60] * 20
}

df = pd.DataFrame(data)
df.head()
```

## 1. Calculate attendance percentage

```python
df['Absent_Days'] = df['Total_Days'] - df['Present_Days']
df['Attendance_Percentage'] = (df['Present_Days'] / df['Total_Days']) * 100
df['Attendance_Percentage'] = df['Attendance_Percentage'].round(2)
df
```

## 2. Attendance summary

```python
print('Number of students:', len(df))
print('Average attendance: {:.2f}%'.format(df['Attendance_Percentage'].mean()))
print('Highest attendance: {:.2f}%'.format(df['Attendance_Percentage'].max()))
print('Lowest attendance: {:.2f}%'.format(df['Attendance_Percentage'].min()))
print('Total present days:', df['Present_Days'].sum())
print('Total absent days:', df['Absent_Days'].sum())
```

## 3. Identify attendance status

For this sample analysis, attendance below 75% is marked as **Short Attendance**. This is an illustrative threshold; follow your institution's actual rules.

```python
df['Status'] = df['Attendance_Percentage'].apply(
    lambda x: 'Short Attendance' if x < 75 else 'Satisfactory'
)
print('Students with short attendance:')
display(df[df['Status'] == 'Short Attendance'][['Student_ID', 'Student_Name', 'Attendance_Percentage', 'Status']])
print('Attendance status count:')
display(df['Status'].value_counts().rename_axis('Status').to_frame('Number_of_Students'))
```

## 4. Bar chart: attendance percentage of each student

```python
plt.figure(figsize=(12, 5))
plt.bar(df['Student_Name'], df['Attendance_Percentage'])
plt.axhline(75, linestyle='--', label='75% reference')
plt.xlabel('Student Name')
plt.ylabel('Attendance Percentage (%)')
plt.title('Attendance Percentage of 20 Students')
plt.xticks(rotation=45)
plt.ylim(0, 105)
plt.legend()
plt.tight_layout()
plt.show()
```

## 5. Pie chart: present days vs absent days (whole class)

```python
total_present = df['Present_Days'].sum()
total_absent = df['Absent_Days'].sum()

plt.figure(figsize=(6, 6))
plt.pie([total_present, total_absent], labels=['Present Days', 'Absent Days'],
        autopct='%1.1f%%', startangle=90)
plt.title('Overall Class Attendance')
plt.show()
```

## 6. Sort students by attendance

```python
ranking = df.sort_values('Attendance_Percentage', ascending=False)
ranking[['Student_ID', 'Student_Name', 'Present_Days', 'Absent_Days', 'Attendance_Percentage', 'Status']]
```

## 7. Conclusion

This analysis calculates each student's attendance percentage, summarizes class attendance, identifies students below the sample 75% threshold, and presents the results using charts. The sample data can be replaced with actual classroom attendance records.
